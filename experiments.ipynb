# Resume RAG Experiments

Use this notebook for assignment analysis: retrieval quality, latency, and hybrid scoring.

```python
import json
import time
import subprocess
from pathlib import Path
```

## 1) Build/refresh resume index

```python
cmd = [
    "python",
    "resume_rag.py",
    "--resume-dir", "data/resumes",
    "--embedding-provider", "huggingface",
    "--embedding-model", "sentence-transformers/all-MiniLM-L6-v2",
]
print(" ".join(cmd))
subprocess.run(cmd, check=True)
```

## 2) Match a job description and capture latency

```python
jd_text = "Need a Python engineer with 5+ years experience in ML, NLP, AWS and Docker." 
start = time.time()
result = subprocess.run([
    "python", "job_matcher.py",
    "--job-description", jd_text,
    "--top-k", "10"
], check=True, capture_output=True, text=True)
latency = time.time() - start
output = json.loads(result.stdout)
print(f"Latency: {latency:.3f}s")
print(json.dumps(output["top_matches"][:3], indent=2))
```
